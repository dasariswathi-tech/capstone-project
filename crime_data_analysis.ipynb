```python
CRIME OCCURANCE OVER TIME:
```

```python
df['Date_Rptd']=pd.to_datetime(df['Date_Rptd'])
df['DATE_OCC']=pd.to_datetime(df['DATE_OCC'])
```

```python
df.info()
```

```python
5.SPATIAL ANALYSIS
#Utilize the geographical information (Latitude and Longitude) to perform spatial analysis.
```

```python
m=df.resample('M',on='Date_Rptd').size()
```

```python
plt.figure(figsize=(10,6))
plt.plot(m.index,m.values,marker='*',color='blue')
plt.title('Crime Occurance over time')
plt.xlabel('month')
plt.ylabel('Crime Count')
plt.grid()
plt.show()
```

```python
 # Visualize crime hotspots on a map.
```

```python
plt.figure(figsize=(12,6))
plt.scatter(df['LON'],df['LAT'],s=14,color='green')
plt.xlabel('Longitude')
plt.ylabel('Latitude')
plt.title('Crime in Geographical conditions')
plt.grid()
plt.show()
```

```python
plt.figure(figsize=(12,6))
sns.lineplot(x=df['LON'],y=df['LAT'],color='blue')
plt.xlabel('Longitude')
plt.ylabel('Latitude')
plt.title('Crime in Geographical conditions')
plt.grid()
plt.show()
```

```python
location_count=df['Location'].value_counts()
print(location_count)
```

```python
#based on crime code
crime_code=df['Crm_Cd']
crime_count=crime_code.value_counts()
print(crime_count)
```

```python
plt.figure(figsize=(10,5))
sns.histplot(df['Crm_Cd'],color='skyblue',edgecolor='black')
plt.title('Most common crimes based on their crime code')
plt.xlabel('Crime Code')
plt.ylabel('count')
plt.grid()
plt.show()
```

```python
6.Victim Demographics:
#Investigate the distribution of victim ages and genders
```

```python
plt.figure(figsize=(12,6))
sns.histplot(df['Vict_Age'],color='yellow',edgecolor='black')
plt.title('Victims count based on their age')
plt.xlabel('Age')
plt.ylabel('count')
plt.grid()
plt.show()
```

```python
#here 0 indicates to the people whose age is not mentioned or not defined in the given data.
```

```python
plt.figure(figsize=(10,5))
sns.countplot(x=df['Vict_Sex'],edgecolor='brown',color='orange')
plt.xlabel('Victim Gender')
plt.title('Distribution of Victim Gender')
plt.grid()
plt.show()
```

```python
#  Identify common premises descriptions where crimes occur.
```

```python
plt.figure(figsize=(10,5))
sns.countplot(x=df['Premis_Desc'],order=df['Premis_Desc'].value_counts().index[:12],palette='pastel')
plt.xlabel('Premises Discription')
plt.ylabel('Frequency')
plt.title('Description of common premises')
plt.xticks(rotation=90)
plt.show()
```

```python
Q=df['Premis_Desc']
count=Q.value_counts()
print(count)
```
